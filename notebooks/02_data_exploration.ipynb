```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from pathlib import Path
from m5_forecasting.data.loader import load_data
from m5_forecasting.data.preprocessing import create_sample
from m5_forecasting.data.merger import melt_sales
from m5_forecasting.data.merger import merge_calendar
from m5_forecasting.data.merger import merge_prices

```

```python
DATA_PATH = Path("../data/raw")
calendar, sales, prices = load_data(DATA_PATH)
```

```python
sales_sample = create_sample(sales, n_products=100)
```

```python
sales_long = melt_sales(sales_sample)
```

```python
sales_calendar = merge_calendar(
    sales_long,
    calendar
)
```

```python
final_data = merge_prices(
    sales_calendar,
    prices
)
```
